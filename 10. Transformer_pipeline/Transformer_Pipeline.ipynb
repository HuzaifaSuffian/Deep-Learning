```python
from transformers import pipeline
```

```python
# Sentiment Analysis
sent_ana = pipeline('sentiment-analysis')
sent_ana('I really like the depth of this course')
```

```python
sent_ana = pipeline('sentiment-analysis')
sent_ana('Many of the topics discussed here are very difficult to understand')
```

```python
# Zero Shot classification: Classifying text without knowing specific label
zsc = pipeline('zero-shot-classification')
zsc('Psychology is one of the difficult subjects to understand',['Education','Business','Politics'])
```

```python
#fill mask: It is used for fill in the blanks
unmasker = pipeline('fill-mask')
unmasker('Football is my <mask> game',top_k=2)
```

```python
# Name Entity Recognition: It recognizes Person, Location, and Organization
ner = pipeline(task='ner',model='NYTK/named-entity-recognition-nerkor-hubert-hungarian',aggregation_strategy='max')
ner('My name is John Doe and I work at Hugging Face in California')
```

```python
# Qna: Here It is not giving answer from itself but from given context
qna = pipeline(task='question-answering')
qna(question='Where do I work', context='My Name is John Doe and I work at Microsoft in San Fransico')
```
