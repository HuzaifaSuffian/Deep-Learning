### Padding: A Padding is a layer like a wrap with is added to each side. It helps us in 2 ways:
1. Detect features on Edges properly.
2. As we move forward in next convolution layer the side of image reduces and the padding helps us prevent that.

```python
import numpy as np


import tensorflow
from tensorflow import keras
from keras.datasets import mnist
from keras import Sequential
from keras.layers import Dense,Conv2D,Flatten
```

```python
(X_train, y_train), (X_test, y_test) = mnist.load_data()
```

```python
X_train[0].shape
```

# Without Padding

```python
model = Sequential()

model.add(Conv2D(32,kernel_size=(3,3), padding='valid',activation='relu',input_shape=(28,28,1)))
model.add(Conv2D(32,kernel_size=(3,3),padding='valid',activation='relu'))

model.add(Flatten())

model.add(Dense(128,activation='relu'))
model.add(Dense(1,activation='sigmoid'))

model.summary()
```

# With Padding

```python
model1 = Sequential()

model1.add(Conv2D(32,kernel_size=(3,3), padding='same',activation='relu',input_shape=(28,28,1)))
model1.add(Conv2D(32,kernel_size=(3,3),padding='same',activation='relu'))

model1.add(Flatten())

model1.add(Dense(128,activation='relu'))
model1.add(Dense(1,activation='sigmoid'))

model1.summary()
```
