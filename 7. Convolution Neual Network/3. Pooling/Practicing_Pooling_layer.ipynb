# Pooling: A pooling is a layer which is added right after Convolution layer. It helps in 2 ways:
## Types of Pooling:
* MaxPooling: Within a specific window of pooling, it takes maximum value
* MinPooling: Within a specific window of pooling, it takes minimum value
* AvgPooling: Within a specific window of pooling, it takes average value
* GlobalMaxPooling: Within a specific window of pooling, it takes maximum value of image value
* GlobalAvgPooling: Within a specific window of pooling, it takes average value of image value

## Benefit
1. Capture high layer feature which helps in faster model and low storage size.
2. Translation invariance: It perform translation invariance which means make a generalize model and is not affect by the position of object in image.

## Disadvantage:
1. Reduce the size of image to half (generally can be customized).
2. Sometimes translation variance is important like self-driving car.


```python
import numpy as np


import tensorflow
from tensorflow import keras
from keras import Sequential
from keras.datasets import mnist
from keras.layers import Conv2D,Flatten,Dense,MaxPooling2D
```

```python
(X_train,y_train),(X_test,y_test) = mnist.load_data()
```

```python
model = Sequential()

model.add(Conv2D(32, kernel_size=(3,3), padding='valid',activation='relu', input_shape=(28,28,1)))
model.add(MaxPooling2D(pool_size=(2,2), strides=(2,2),padding='valid'))

model.add(Conv2D(32, kernel_size=(3,3), padding='valid',activation='relu'))
model.add(MaxPooling2D(pool_size=(2,2), strides=(2,2),padding='valid'))

model.add(Flatten())

model.add(Dense(32, activation='relu'))
model.add(Dense(10, activation='softmax'))

model.summary()
```
